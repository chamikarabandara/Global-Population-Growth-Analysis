# STEP 6 — Machine Learning Data Preparation

## Objective

The objective of this step is to prepare the feature-engineered population dataset for machine learning.

### Main Tasks

- Load the feature-engineered dataset
- Select the input features
- Define the target variable
- Split the data into training and testing sets using a time-based split

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "../data/processed/Global_Population_Feature_Engineered.csv"
)

print("Dataset Shape:", df.shape)
df.head()
```

```python
print("Minimum Year:", df["Year"].min())
print("Maximum Year:", df["Year"].max())
print("Number of Countries:", df["Country Code"].nunique())
```

```python
features = [
    "Year",
    "Previous_Year_Population",
    "Population_Growth_Percentage",
    "Population_Lag_2",
    "Population_Rolling_Avg_3"
]

target = "Population"

X = df[features]
y = df[target]

print("Features:")
print(features)

print("\nTarget:")
print(target)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
train_df = df[df["Year"] <= 2020].copy()
test_df = df[df["Year"] > 2020].copy()

print("Training Data:", train_df.shape)
print("Testing Data:", test_df.shape)

print("\nTraining Years:",
      train_df["Year"].min(),
      "-",
      train_df["Year"].max())

print("Testing Years:",
      test_df["Year"].min(),
      "-",
      test_df["Year"].max())
```

```python
X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

## Data Preparation Summary

The feature-engineered population dataset was divided into training and testing datasets using a time-based split.

### Training Dataset
- Period: 1962–2020
- Records: 12,773
- Features: 5

### Testing Dataset
- Period: 2021–2025
- Records: 1,085
- Features: 5

### Input Features

1. Year
2. Previous Year Population
3. Population Growth Percentage
4. Population Lag 2
5. Population Rolling Average (3 years)

### Target Variable

- Population

A time-based split was used to prevent future information from being used during model training.
