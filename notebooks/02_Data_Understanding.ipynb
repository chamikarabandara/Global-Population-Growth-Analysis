## 2. Data Understanding

This notebook focuses on understanding the World Bank population dataset before data cleaning and exploratory data analysis.

The dataset contains population information for countries and other World Bank entities over multiple years.

```python
import pandas as pd
from pathlib import Path

print("Libraries imported successfully!")
```

```python
BASE_DIR = Path.cwd().parent

RAW_DATA_PATH = (
    BASE_DIR
    / "data"
    / "raw"
    / "API_SP.POP.TOTL_DS2_en_csv_v2_350115.csv"
)

df = pd.read_csv(RAW_DATA_PATH, skiprows=4)

print("Raw dataset loaded successfully!")
```

```python
print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])
print("Dataset shape     :", df.shape)
```

```python
print("Total World Bank entities:", df["Country Name"].nunique())
```

```python
print("First 30 entities:\n")

print(df["Country Name"].unique()[:30])
```

```python
year_columns = [
    column for column in df.columns
    if str(column).isdigit()
]

print("First year:", year_columns[0])
print("Last year :", year_columns[-1])
print("Number of years:", len(year_columns))
```

```python
missing_values = df.isnull().sum()

print("Missing values by column:\n")
print(missing_values[missing_values > 0])
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
print("Indicator Names:\n")
print(df["Indicator Name"].unique())
```

```python
print("Number of unique country/entity codes:",
      df["Country Code"].nunique())
```
