# STEP 5 — Feature Engineering

## Objective

The objective of this step is to create additional features from the cleaned population dataset for machine learning.

### Features to be Created

- Previous Year Population
- Population Growth Percentage
- Population Lag 2
- 3-Year Rolling Average Population

These features will help machine learning models learn population trends and historical patterns.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "../data/processed/Global_Population_Cleaned.csv"
)

df.head()
```

```python
df = df.sort_values(
    ["Country Code", "Year"]
).reset_index(drop=True)

df.head()
```

```python
df["Previous_Year_Population"] = (
    df.groupby("Country Code")["Population"]
      .shift(1)
)

df.head(10)
```

Population Growth %

```python
df["Population_Growth_Percentage"] = (
    df.groupby("Country Code")["Population"]
      .pct_change() * 100
)

df.head(10)
```

Population Lag 2

```python
df["Population_Lag_2"] = (
    df.groupby("Country Code")["Population"]
      .shift(2)
)

df.head(10)
```

3-Year Rolling Average

```python
df["Population_Rolling_Avg_3"] = (
    df.groupby("Country Code")["Population"]
      .transform(
          lambda x: x.rolling(window=3, min_periods=1).mean()
      )
)

df.head(10)
```

```python
df.head(10)
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
ml_df = df.dropna(
    subset=[
        "Previous_Year_Population",
        "Population_Lag_2"
    ]
).copy()

ml_df.shape
```

```python
output_path = "../data/processed/Global_Population_Feature_Engineered.csv"

ml_df.to_csv(
    output_path,
    index=False
)

print("Feature engineered dataset saved successfully.")
print(output_path)
```
